## The Illusion of memory

```python
from openai import OpenAI

client = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
```

```python
messages = [
    {"role":"system", "content":"You are a helpful assistant."},
    {"role":"user", "content":"Hi! I am Ut!"}
]
```

```python
response = client.chat.completions.create(
    model="qwen2.5-coder:1.5b",
    reasoning_effort="none",
    messages=messages
)


response.choices[0].message.content
```

```python
messages = [
    {"role":"system", "content":"You are a helpful assistant."},
    {"role":"user", "content":"What is my name?"}
]
```

```python
response = client.chat.completions.create(
    model="qwen2.5-coder:1.5b",
    reasoning_effort="none",
    messages=messages
)

response.choices[0].message.content
```

### Wait, wha??

What's going on??

Here's the thing: every call to an LLM is completely STATELESS. It's a totally new call, every single time. As AI engineers, it's OUR JOB to devise techniques to give the impression that the LLM has a "memory".

```python
messages = [
    {"role":"system", "content":"You are a helpful assistant."},
    {"role":"user", "content":"Hi! I am Ut!"},
    {"role":"assistant", "content":"Hello Ut! How can I assist you today?"},
    {"role":"user", "content":"What is my name?"}
]
```

```python
response = client.chat.completions.create(
    model="qwen2.5-coder:1.5b",
    reasoning_effort="none",
    messages=messages
)

response.choices[0].message.content
```

### Summmary

1. Every call to an LLM is stateless
2. We pass in the entire conversation so far in the input prompt, every time
3. This gives the illusion that the LLM has memory - it apparently keeps the context of the conversation
4. But this is a trick; it's a by-product of providing the entire conversation, every time
5. An LLM just predicts the most likely next tokens in the sequence; if that sequence contains "My name is Ut" and later "What's my name?" then it will predict.. Ut!

The ChatGPT product uses exactly this trick - every time you send a message, it's the entire conversation that gets passed in.

"Does that mean we have to pay extra each time for all the conversation so far"

Absolutely. And that's exactly what we want! We want the LLM to look back at the entire conversation and predict the next token based on all the context so far. That requires computation every time we generate a response, which means we have to pay for it. In other words, every bit of context the model processes comes with a computational cost—and someone has to pay the electricity bill! 😄
