## Tokenization

```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")

tokens = encoding.encode("Hello I start learning LLMs!")
```

```python
tokens
```

```python
for token_id in tokens:
    token_text = encoding.decode([token_id])
    print(f"{token_text} = {token_id}")
```
